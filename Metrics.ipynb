```python
import os
import csv
import pandas as pd

def Highest_training_metrics(df):
    # Get the index of the row with the highest Training Accuracy
    highest_accuracy_idx = df['Training Accuracy'].idxmax()
    # Get the entire row with the highest Training Accuracy
    highest_accuracy_row = df.loc[highest_accuracy_idx]
    return highest_accuracy_row

def Highest_validation_metrics(df):
    # Get the index of the row with the highest Validation Accuracy
    highest_accuracy_idx = df['Validation Accuracy'].idxmax()
    # Get the entire row with the highest Validation Accuracy
    highest_accuracy_row = df.loc[highest_accuracy_idx]
    return highest_accuracy_row

folders = [
    r"C:\Users\Tharunraj_S\Desktop\Major project\codes\Pytorch\MFCC\N_MFCC\CNN_LSTM_Model",
    r"C:\Users\Tharunraj_S\Desktop\Major project\codes\Pytorch\MFCC\N_MFCC\LSTMModel",
    r"C:\Users\Tharunraj_S\Desktop\Major project\codes\Pytorch\MFCC\N_MFCC\Resnet18_LSTM_Model"
]

model_names = ['CNN_LSTM', 'LSTM','Resnet18_LSTM_Model']

datasets = ['tess', 'crema', 'savee', 'ravdess']
levels = [16, 24, 32, 40]
mfccs = ['N_MFCC']
epochs = [500]

with open('Training_Highest_Accuracy_Metrics.txt', 'w') as f_train_txt, open('Training_Highest_Accuracy_Metrics.csv', 'w', newline='') as f_train_csv:
    train_csv_writer = csv.writer(f_train_csv)
    train_csv_writer.writerow(['Model', 'Dataset', 'NMFCC Level', 'Total Epochs', 'Training Loss', 'Training Accuracy', 'Training Precision', 'Training Recall', 'Training F1-Score'])  # Writing header for training CSV file

    for folder, model_name in zip(folders, model_names):
        for dataset in datasets:
            for level in levels:
                for mfcc in mfccs:
                    for epoch in epochs:
                        variable_name = f"{dataset}{level}_{epoch}"
                        file_path = os.path.join(folder, dataset, f"level {level} wavelet", f"{epoch} Epochs", "metrics.csv")
                        try:
                            df = pd.read_csv(file_path)
                            if 'Training Loss' in df.columns:
                                # Get the row with the highest training accuracy
                                highest_train_accuracy_row = Highest_training_metrics(df)
                                # Writing to training CSV file
                                train_csv_writer.writerow([model_name, dataset, level, epoch] + highest_train_accuracy_row[['Training Loss', 'Training Accuracy', 'Training Precision', 'Training Recall', 'Training F1-Score']].values.tolist())
                                # Writing to training text file
                                f_train_txt.write(f"Model: {model_name}, Dataset: {dataset}, Level: {level}, Epochs: {epoch}\n")
                                f_train_txt.write(str(highest_train_accuracy_row[['Training Loss', 'Training Accuracy', 'Training Precision', 'Training Recall', 'Training F1-Score']]) + "\n")
                                f_train_txt.write("="*50 + "\n")  # Separating different dataframes
                        except FileNotFoundError:
                            print(f"Error: File '{file_path}' not found.")

with open('Validation_Highest_Accuracy_Metrics.txt', 'w') as f_val_txt, open('Validation_Highest_Accuracy_Metrics.csv', 'w', newline='') as f_val_csv:
    val_csv_writer = csv.writer(f_val_csv)
    val_csv_writer.writerow(['Model', 'Dataset', 'NMFCC Level', 'Total Epochs', 'Validation Loss', 'Validation Accuracy', 'Validation Precision', 'Validation Recall', 'Validation F1-Score'])  # Writing header for validation CSV file

    for folder, model_name in zip(folders, model_names):
        for dataset in datasets:
            for level in levels:
                for mfcc in mfccs:
                    for epoch in epochs:
                        variable_name = f"{dataset}{level}_{epoch}"
                        file_path = os.path.join(folder, dataset, f"level {level} wavelet", f"{epoch} Epochs", "metrics.csv")
                        try:
                            df = pd.read_csv(file_path)
                            if 'Validation Loss' in df.columns:
                                # Get the row with the highest validation accuracy
                                highest_val_accuracy_row = Highest_validation_metrics(df)
                                # Writing to validation CSV file
                                val_csv_writer.writerow([model_name, dataset, level, epoch] + highest_val_accuracy_row[['Validation Loss', 'Validation Accuracy', 'Validatio Precision', 'Validatio Recall', 'Validatio F1-Score']].values.tolist())
                                # Writing to validation text file
                                f_val_txt.write(f"Model: {model_name}, Dataset: {dataset}, Level: {level}, Epochs: {epoch}\n")
                                f_val_txt.write(str(highest_val_accuracy_row[['Validation Loss', 'Validation Accuracy', 'Validatio Precision', 'Validatio Recall', 'Validatio F1-Score']]) + "\n")
                                f_val_txt.write("="*50 + "\n")  # Separating different dataframes
                        except FileNotFoundError:
                            print(f"Error: File '{file_path}' not found.")
```
