```python
import os
import csv
import pandas as pd

def highest_training_metrics(df):
    highest_accuracy_idx = df['Training Accuracy'].idxmax()
    highest_accuracy_row = df.loc[highest_accuracy_idx]
    return highest_accuracy_row

def highest_validation_metrics(df):
    highest_accuracy_idx = df['Validation Accuracy'].idxmax()
    highest_accuracy_row = df.loc[highest_accuracy_idx]
    return highest_accuracy_row

root_folders = [
    r"C:\Users\Tharunraj_S\Desktop\Major project\codes\Pytorch\DWT and MFCC\CNN_LSTM_Model",
    r"C:\Users\Tharunraj_S\Desktop\Major project\codes\Pytorch\DWT and MFCC\LSTMModel"
]
model_names = ['CNN_LSTM', 'LSTM']
datasets = ['tess', 'crema', 'savee', 'ravdess']
levels = [1, 2, 3, 4, 5]
mfccs = ['N_MFCC16', 'N_MFCC24', 'N_MFCC32', 'N_MFCC40']
epochs = [500]

for root_folder, model_name in zip(root_folders, model_names):
    best_train_metrics = {dataset: {'accuracy': -1} for dataset in datasets}
    best_val_metrics = {dataset: {'accuracy': -1} for dataset in datasets}
    
    for dataset in datasets:
        for mfcc in mfccs:
            for level in levels:
                for epoch in epochs:
                    file_path = os.path.join(root_folder, dataset, mfcc, f"level {level} wavelet", f"{epoch} Epochs", "metrics.csv")
                    if os.path.exists(file_path):
                        df = pd.read_csv(file_path)
                        if 'Training Loss' in df.columns:
                            highest_train_accuracy_row = highest_training_metrics(df)
                            if highest_train_accuracy_row['Training Accuracy'] > best_train_metrics[dataset]['accuracy']:
                                best_train_metrics[dataset] = {
                                    'accuracy': highest_train_accuracy_row['Training Accuracy'],
                                    'row': highest_train_accuracy_row,
                                    'mfcc': mfcc[-2:],
                                    'dwt_level': level,
                                    'epochs': epoch
                                }
                        if 'Validation Loss' in df.columns:
                            highest_val_accuracy_row = highest_validation_metrics(df)
                            if highest_val_accuracy_row['Validation Accuracy'] > best_val_metrics[dataset]['accuracy']:
                                best_val_metrics[dataset] = {
                                    'accuracy': highest_val_accuracy_row['Validation Accuracy'],
                                    'row': highest_val_accuracy_row,
                                    'mfcc': mfcc[-2:],
                                    'dwt_level': level,
                                    'epochs': epoch
                                }

    # Write the best training metrics to file
    with open(f'Training_Highest_Accuracy_Metrics_{model_name}.txt', 'w') as f_train_txt, open(f'Training_Highest_Accuracy_Metrics_{model_name}.csv', 'w', newline='') as f_train_csv:
        train_csv_writer = csv.writer(f_train_csv)
        train_csv_writer.writerow(['Model', 'Dataset', 'NMFCC Level', 'DWT Level', 'Total Epochs', 'Training Loss', 'Training Accuracy', 'Training Precision', 'Training Recall', 'Training F1-Score'])

        for dataset in datasets:
            best_train_row = best_train_metrics[dataset]['row']
            train_csv_writer.writerow([model_name, dataset, best_train_metrics[dataset]['mfcc'], best_train_metrics[dataset]['dwt_level'], best_train_metrics[dataset]['epochs']] + best_train_row[['Training Loss', 'Training Accuracy', 'Training Precision', 'Training Recall', 'Training F1-Score']].values.tolist())
            f_train_txt.write(f"Model: {model_name}, Dataset: {dataset}, MFCC: {best_train_metrics[dataset]['mfcc']}, DWT Level: {best_train_metrics[dataset]['dwt_level']}, Epochs: {best_train_metrics[dataset]['epochs']}\n")
            f_train_txt.write(str(best_train_row[['Training Loss', 'Training Accuracy', 'Training Precision', 'Training Recall', 'Training F1-Score']]) + "\n")
            f_train_txt.write("="*50 + "\n")

    # Write the best validation metrics to file
    with open(f'Validation_Highest_Accuracy_Metrics_{model_name}.txt', 'w') as f_val_txt, open(f'Validation_Highest_Accuracy_Metrics_{model_name}.csv', 'w', newline='') as f_val_csv:
        val_csv_writer = csv.writer(f_val_csv)
        val_csv_writer.writerow(['Model', 'Dataset', 'NMFCC Level', 'DWT Level', 'Total Epochs', 'Validation Loss', 'Validation Accuracy', 'Validation Precision', 'Validation Recall', 'Validation F1-Score'])

        for dataset in datasets:
            best_val_row = best_val_metrics[dataset]['row']
            val_csv_writer.writerow([model_name, dataset, best_val_metrics[dataset]['mfcc'], best_val_metrics[dataset]['dwt_level'], best_val_metrics[dataset]['epochs']] + best_val_row[['Validation Loss', 'Validation Accuracy', 'Validation Precision', 'Validation Recall', 'Validation F1-Score']].values.tolist())
            f_val_txt.write(f"Model: {model_name}, Dataset: {dataset}, MFCC: {best_val_metrics[dataset]['mfcc']}, DWT Level: {best_val_metrics[dataset]['dwt_level']}, Epochs: {best_val_metrics[dataset]['epochs']}\n")
            f_val_txt.write(str(best_val_row[['Validation Loss', 'Validation Accuracy', 'Validation Precision', 'Validation Recall', 'Validation F1-Score']]) + "\n")
            f_val_txt.write("="*50 + "\n")
```
