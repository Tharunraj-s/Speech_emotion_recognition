# LSTM GRAPH

```python
import pandas as pd
import matplotlib.pyplot as plt

# Step 1: Read CSV file
df = pd.read_csv(r"DWT and MFCC\DWT level experiment\LSTMModel\tess\level 4 wavelet\500 Epochs\metrics.csv")
```

```python
# Step 2: Extract necessary data
epochs = df['Epoch']
training_loss = df['Training Loss']
training_accuracy = df['Training Accuracy']
training_precision = df['Training Precision']
training_recall = df['Training Recall']
training_f1_score = df['Training F1-Score']

validation_loss = df['Validation Loss']
validation_accuracy = df['Validation Accuracy']
validation_precision = df['Validation Precision']  
validation_recall = df['Validation Recall']  
validation_f1_score = df['Validation F1-Score']  

# Step 3: Apply moving averages
window_size = 3  # You can adjust this window size as needed
training_loss_smooth = training_loss.rolling(window=window_size).mean()
validation_loss_smooth = validation_loss.rolling(window=window_size).mean()

training_accuracy_smooth = training_accuracy.rolling(window=window_size).mean()
validation_accuracy_smooth = validation_accuracy.rolling(window=window_size).mean()

training_precision_smooth = training_precision.rolling(window=window_size).mean()
validation_precision_smooth = validation_precision.rolling(window=window_size).mean()

training_recall_smooth = training_recall.rolling(window=window_size).mean()
validation_recall_smooth = validation_recall.rolling(window=window_size).mean()

# Step 4: Plotting
plt.figure(figsize=(12, 8))
# Training and Validation Loss
plt.subplot(2, 2, 1)
plt.plot(epochs, training_loss_smooth, label='Training Loss', linewidth=2)
plt.plot(epochs, validation_loss_smooth, label='Validation Loss', linewidth=2)
plt.title('LSTM Loss', fontsize=16)
plt.xlabel('Epochs', fontsize=14)
plt.ylabel('Loss', fontsize=14)
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
plt.legend(fontsize=12)

# Training and Validation Accuracy
plt.subplot(2, 2, 2)
plt.plot(epochs, training_accuracy_smooth, label='Training Accuracy', linewidth=2)
plt.plot(epochs, validation_accuracy_smooth, label='Validation Accuracy', linewidth=2)
plt.title('LSTM Accuracy', fontsize=16)
plt.xlabel('Epochs', fontsize=14)
plt.ylabel('Accuracy', fontsize=14)
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
plt.legend(fontsize=12)

# Training and Validation Precision
plt.subplot(2, 2, 3)
plt.plot(epochs, training_precision_smooth, label='Training Precision', linewidth=2)
plt.plot(epochs, validation_precision_smooth, label='Validation Precision', linewidth=2)
plt.title('LSTM Precision', fontsize=16)
plt.xlabel('Epochs', fontsize=14)
plt.ylabel('Precision', fontsize=14)
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
plt.legend(fontsize=12)

# Training and Validation Recall
plt.subplot(2, 2, 4)
plt.plot(epochs, training_recall_smooth, label='Training Recall', linewidth=2)
plt.plot(epochs, validation_recall_smooth, label='Validation Recall', linewidth=2)
plt.title('LSTM Recall', fontsize=16)
plt.xlabel('Epochs', fontsize=14)
plt.ylabel('Recall', fontsize=14)
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
plt.legend(fontsize=12)

plt.tight_layout()
plt.savefig('LSTM Graphs.png')
plt.show()
```
